```python
import os
import numpy as np
import pandas as pd
import lightkurve as lk
from lightkurve import LightCurveCollection
from tqdm import tqdm
from sklearn.utils import shuffle, resample
from typing import List, Optional
from scipy.signal import savgol_filter
from requests.exceptions import ReadTimeout, ConnectionError, ChunkedEncodingError
import time
```

```python
all_df = pd.read_csv("../data/all_df.csv")
```

```python
def safe_download(product, retries=3, timeout=60, delay=5):
    """
    Safely download a single light curve product with retries and timeout.
    """
    for attempt in range(retries):
        try:
            return product.download(timeout=timeout)
        except (ReadTimeout, ConnectionError, ChunkedEncodingError, OSError) as e:
            print(f"[WARN] Download failed ({e}). Retry {attempt+1}/{retries} in {delay}s...")
            time.sleep(delay)
    print("[SKIP] Giving up on this file after multiple retries.")
    return None
```

```python
def balance_catalog(df, catalog, n_samples=None):
    df = df[df["catalog"] == catalog]
    df = df[df["period"].notnull() & df["epoch"].notnull()]
    
    positives = df[df["label"]==1]
    negatives = df[df["label"]==0]

    n = min(len(positives), len(negatives))
    if n_samples:
        n = min(n, n_samples)
    
    pos = positives.sample(n, random_state=42)
    neg = negatives.sample(n, random_state=42)
    
    return pd.concat([pos, neg]).sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
def process_lightcurve(target_id, mission, period, epoch,
                       n_global=2001, n_local=201, local_window=0.05,
                       fast_mode=True, show_progress=False):
    """
    Download + process one light curve (with error handling).
    If a star fails repeatedly, it is skipped so the pipeline keeps running.
    """
    try:
        search = lk.search_lightcurve(target_id, mission=mission)
        if len(search) == 0:
            print(f"[SKIP] {target_id}: no data products")
            return None, None

        if fast_mode:
            lc = safe_download(search[0], retries=3, timeout=60)
            if lc is None:
                return None, None
            lc = lc.remove_nans().flatten()
        else:
            lcs = []
            # Quarter/sector loop with error handling
            for i, product in enumerate(search, start=1):
                print(f"   {target_id}: downloading {i}/{len(search)}...")
                dl = safe_download(product, retries=3, timeout=90)
                if dl is not None:
                    lcs.append(dl)
            if not lcs:
                print(f"[SKIP] {target_id}: all downloads failed")
                return None, None
            lc = lcs[0] if len(lcs)==1 else LightCurveCollection(lcs).stitch()
            lc = lc.remove_nans().flatten()

        # Ephemeris sanity check
        if period is None or np.isnan(period) or epoch is None or np.isnan(epoch):
            print(f"[SKIP] {target_id}: missing/invalid period/epoch")
            return None, None

        # Fold
        folded = lc.fold(period=period, epoch_time=epoch)
        phase, flux = folded.phase.value, folded.flux.value
        flux = (flux - np.nanmedian(flux)) / np.nanstd(flux)
        order = np.argsort(phase)
        phase, flux = phase[order], flux[order]

        # Global view
        global_phase = np.linspace(-0.5, 0.5, n_global)
        global_flux  = np.interp(global_phase, phase, flux)

        # Local view
        mask = (phase > -local_window) & (phase < local_window)
        if np.sum(mask) < 3:
            print(f"[SKIP] {target_id}: too few transit points")
            return None, None
        local_phase = np.linspace(-local_window, local_window, n_local)
        local_flux  = np.interp(local_phase, phase[mask], flux[mask])

        print(f"[OK] {target_id}: processed successfully.")
        return global_flux, local_flux

    except Exception as e:
        print(f"[FAIL] {target_id}: unexpected error {e}")
        return None, None
```

```python
def build_dataset(catalog_df, mission, prefix,
                  fast_mode=True, show_progress=True):
    """
    Process a catalog (KOI or TOI) into global/local arrays with progress bar
    and final summary counts.
    """
    Xg, Xl, y = [], [], []
    successes, skips = 0, 0

    for _, row in tqdm(catalog_df.iterrows(), total=len(catalog_df),
                       desc=f"Processing {mission} targets", disable=not show_progress):
        target = f"{prefix} {int(row['target_id'])}"
        g, l = process_lightcurve(target, mission, row["period"], row["epoch"],
                                  fast_mode=fast_mode)
        if g is not None and l is not None:
            Xg.append(g); Xl.append(l); y.append(row["label"])
            successes += 1
        else:
            skips += 1

    # Convert arrays
    Xg, Xl, y = np.array(Xg, np.float32), np.array(Xl, np.float32), np.array(y, np.int8)

    # ✅ Summary
    total = len(catalog_df)
    print(f"\n[SUMMARY] {mission}:")
    print(f"   Total targets:    {total}")
    print(f"   Processed OK:     {successes}")
    print(f"   Skipped (failed): {skips}")
    print(f"   Yield:            {100*successes/total:.1f}%\n")

    return Xg, Xl, y
```

```python
def build_koi_toi_dataset(all_df, n_samples=None, fast_mode=True):
    koi_balanced = balance_catalog(all_df, "KOI", n_samples=2000)
    toi_balanced = balance_catalog(all_df, "TOI", n_samples=2000)

    print("[INFO] Building KOI dataset...")
    Xg_koi, Xl_koi, y_koi = build_dataset(koi_balanced, "Kepler", "KIC", fast_mode=True)

    print("[INFO] Building TOI dataset...")
    Xg_toi, Xl_toi, y_toi = build_dataset(toi_balanced, "TESS", "TIC", fast_mode=True)

    # Merge
    Xg = np.vstack([Xg_koi, Xg_toi])
    Xl = np.vstack([Xl_koi, Xl_toi])
    y  = np.hstack([y_koi, y_toi])

    print(f"[FINAL SUMMARY] Overall dataset: {Xg.shape[0]} usable samples")
    np.savez("KOI_TOI_dataset.npz", X_global=Xg, X_local=Xl, y=y)

    return Xg, Xl, y
```

```python
Xg, Xl, y = build_koi_toi_dataset(all_df, n_samples=2000, fast_mode=False)
```
