```python
import pandas as pd
import numpy as np
import xgboost as xgb
from tqdm import tqdm
import lightkurve as lk
```

```python
def process_lightcurve(target_id, mission="Kepler", period=None, epoch=None,
                       n_global=2001, n_local=201, local_window=0.05):
    """
    Download and process a light curve -> detrend, fold, global & local arrays.
    
    Parameters
    ----------
    target_id : str or int
        KIC or TIC identifier.
    mission : str
        'Kepler' or 'TESS'.
    period : float
        Transit period [days].
    epoch : float
        Transit epoch (t0).
    n_global : int
        Length of global array.
    n_local : int
        Length of local array.
    local_window : float
        Phase size to extract around transit (e.g. +/-0.05).
        
    Returns
    -------
    global_array, local_array : np.ndarray
        Phase-folded flux arrays ready for ML.
    """
    try:
        search = lk.search_lightcurve(target_id, mission=mission)
        lc = search.download_all().stitch().remove_nans()

        lc = lc.flatten()

        if period is None or epoch is None:
            return None, None

        fold_lc = lc.fold(period=period, epoch_time=epoch)

        phase = fold_lc.phase.value
        flux = fold_lc.flux.value
        flux = (flux - np.nanmedian(flux)) / np.nanstd(flux)

        order = np.argsort(phase)
        phase = phase[order]
        flux = flux[order]

        global_phase = np.linspace(-0.5, 0.5, n_global)
        global_flux = np.interp(global_phase, phase, flux)

        mask = (phase > -local_window) & (phase < local_window)
        local_phase = np.linspace(-local_window, local_window, n_local)
        local_flux = np.interp(local_phase, phase[mask], flux[mask])

        return global_flux, local_flux
    
    except Exception as e:
        print(f"Failed for target {target_id}: {e}")
        return None, None
```

```python
def build_and_save_dataset(catalog_df, mission="Kepler",
                           out_file="ml_dataset.npz", n_samples=100):
    """
    Build dataset from catalog_df -> (X_global, X_local, y)
    Saves result into an .npz file.
    """
    Xg, Xl, y = [], [], []

    sampled = catalog_df.sample(n=n_samples, random_state=42)

    for _, row in tqdm(sampled.iterrows(), total=len(sampled)):
        tid = row["target_id"]
        period, epoch, label = row["period"], row["epoch"], row["label"]

        if pd.isna(period) or pd.isna(epoch):
            continue

        # Proper identifier format: KIC NNN / TIC NNN
        if mission == "Kepler":
            target = f"KIC {int(tid)}"
        else:
            target = f"TIC {int(tid)}"

        g, l = process_lightcurve(target, mission=mission,
                                  period=period, epoch=epoch)

        if g is not None and l is not None:
            Xg.append(g); Xl.append(l); y.append(label)

    # Convert to numpy arrays
    Xg, Xl, y = np.array(Xg), np.array(Xl), np.array(y)

    # Save as .npz
    np.savez(out_file, X_global=Xg, X_local=Xl, y=y)
    print(f"Saved dataset: {out_file} with {len(y)} samples")
    return Xg, Xl, y
```

```python
all_df = pd.read_csv("../data/all_df.csv")
koi_df = pd.read_csv("../data/koi_df.csv")
toi_df = pd.read_csv("../data/toi_df.csv")
tce_df = pd.read_csv("../data/tce_df.csv")
```

```python
def get_numeric_features(all_df):
    """
    Select a clean subset of numeric metadata features from merged exoplanet catalogs.
    """
    candidate_cols = [
        "period", "epoch",
        "koi_duration", "koi_depth", "koi_snr",
        "tce_duration", "tce_depth", "tce_mes",
        "toi_duration", "toi_depth",
        "st_rad", "st_mass", "st_teff"  
    ]
    
    available_cols = [c for c in candidate_cols if c in all_df.columns]
    features = all_df[["target_id", "catalog", "label"] + available_cols].copy()
    
    for c in available_cols:
        features[c] = pd.to_numeric(features[c], errors="coerce")
    
    for c in available_cols:
        features[c] = features[c].fillna(features[c].median())
    
    return features
```

```python
valid = all_df[all_df["period"].notnull() & all_df["epoch"].notnull()]

positives = valid[(valid["catalog"]=="KOI") & (valid["label"]==1)]
negatives = valid[(valid["catalog"]=="KOI") & (valid["label"]==0)].sample(len(positives))

train_df = pd.concat([positives, negatives])
```

```python
Xg, Xl, y = build_and_save_dataset(train_df,
                                   mission="Kepler",
                                   out_file="koi_dataset.npz",
                                   n_samples=10)
```
