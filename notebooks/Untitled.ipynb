```python
from astropy.utils.data import get_cache_dir, get_cached_urls, clear_download_cache

# Location of the astroquery/astropy cache
cache_dir = get_cache_dir()
print(f"Astroquery cache directory: {cache_dir}")

# List cached files (URLs mapped to files)
cached_files = list(get_cached_urls())
print(f"Number of cached files: {len(cached_files)}")

# Show a few
for i, url in enumerate(cached_files[:10]):
    print(f"{i+1}: {url}")
```
